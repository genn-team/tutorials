and then copy wheel file into folder with tutorials

```python
!pip install wheel
from platform import python_implementation, python_version_tuple
from shutil import copy
from wheel.bdist_wheel import get_abi_tag, get_platform

assert python_implementation() == "CPython"
version_tuple = python_version_tuple()
wheel_name = f"pyfenn-0.0.1-cp{version_tuple[0]}{version_tuple[1]}-{get_abi_tag()}-{get_platform(None)}.whl"
print(wheel_name)
copy(f"/content/riscv_ise/{wheel_name}", f"/content/drive/MyDrive/genn_tutorials/{wheel_name}")
```
